#### PARTIE 1 - IMPORTS ET CHARGEMENT DES DONNÉES PRÉTRAITÉES


```python
import os
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
# Charger les datasets prétraités
train = pd.read_csv("../data/processed/train_preprocessed.csv")
test = pd.read_csv("../data/processed/test_preprocessed.csv")
```

#### PARTIE 2 - GÉNÉRATION DES EMBEDDINGS

```python
# Charger le modèle pré-entraîné pour transformer le texte en vecteurs
model = SentenceTransformer('all-MiniLM-L6-v2')

# Générer les embeddings pour train et test
train_embeddings = model.encode(train['clean_text'].tolist(), show_progress_bar=True)
test_embeddings = model.encode(test['clean_text'].tolist(), show_progress_bar=True)

# Vérifier la dimension des embeddings
print("Taille des embeddings train :", train_embeddings.shape)
print("Taille des embeddings test :", test_embeddings.shape)
```

#### PARTIE 3 - SAUVEGARDE DES EMBEDDINGS

```python
# Sauvegarder les embeddings sous forme de fichiers .npy
np.save("../data/processed/train_embeddings.npy", train_embeddings)
np.save("../data/processed/test_embeddings.npy", test_embeddings)
```

```python
train= np.load("../data/processed/train_embeddings.npy")
test= np.load("../data/processed/test_embeddings.npy")
print("Forme du train :", train.shape)
print("Forme du test :", test.shape)

# Affiche les 2 premiers vecteurs (10 premières valeurs de chacun)
print("\nExemple du 1er embedding (train):", train[0][:10])
print("\nExemple du 2e embedding (train):", train[1][:10])
```
